```python
#zero-shot classification就是这些你规定的新类别不用 再根据它们去微调模型，
#但本质上模型其实之前预训练时候也训练过这些类别，如果没训练过那么效果会变差
```

```python
#执行下zero-shot classification
from transformers import pipeline



#加载classifier
classifier1 = pipeline("zero-shot-classification")
classifier1(
    "This is a course about the Transformers library",
    candidate_labels=['art', 'human', 'politics'],
)
```

```python
#换一个classifier
classifier2 = pipeline("zero-shot-classification")
classifier2(
    "This is a course about the Transformers library",
    candidate_labels=['education', 'human', 'politics'],
)
```
